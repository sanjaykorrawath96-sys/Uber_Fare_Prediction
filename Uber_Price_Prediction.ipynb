### Data preprocessing

```python
df.isna().sum()
```

```python
# 1. Convert datetime strings to pandas datetime format
df['tpep_pickup_datetime'] = pd.to_datetime(df['tpep_pickup_datetime'])
df['tpep_dropoff_datetime'] = pd.to_datetime(df['tpep_dropoff_datetime'])

# 2. Extract Trip Duration in Minutes
df['duration_minutes'] = (df['tpep_dropoff_datetime'] - df['tpep_pickup_datetime']).dt.total_seconds() / 60.0

# 3. Extract Day of Month (1 - 31)
df['day'] = df['tpep_pickup_datetime'].dt.day

# 4. Extract Day of Week (0 = Monday, 6 = Sunday)
df['day_of_week'] = df['tpep_pickup_datetime'].dt.dayofweek

# 5. Extract Is Weekend flag (1 if Saturday/Sunday, else 0)
df['is_weekend'] = df['day_of_week'].isin([5, 6]).astype(int)

# 6. Extract Is Night flag (1 if pickup between 8 PM and 6 AM, else 0)
df['hour'] = df['tpep_pickup_datetime'].dt.hour
df['is_night'] = ((df['hour'] >= 20) | (df['hour'] < 6)).astype(int)

```

Column	         Why
+ total_amount:	Leakage. It is the fare plus tip, tolls and surcharges.
+ tip_amount:	Leakage. It is only known after the fare is paid, and it is not part of the fare.
+ tolls_amount:	Only known after the trip. I used it in the earlier check, but drop it if you want a pre-trip prediction.
+ mta_tax:	Nearly constant, so it carries no signal.
+ improvement_surcharge:	Nearly constant.
+ extra:	A rush-hour/overnight surcharge. You can rebuild it from hour, so keep hour and drop this.
+ payment_type:	It relates to tips, not to the metered fare.
+ store_and_fwd_flag:	A technical flag about whether the trip record was stored before upload. It has no link to the fare.
+ VendorID:	Low information.
+ tpep_pickup_datetime, tpep_dropoff_datetime:	Drop the raw strings after extracting features from them (see below).

```python
cols_to_drop = [
    "total_amount",
    "tip_amount",
    "tolls_amount",
    "mta_tax",
    "improvement_surcharge",
    "extra",
    "payment_type",
    "store_and_fwd_flag",
    "VendorID",
    "tpep_pickup_datetime",
    "tpep_dropoff_datetime",
    'pickup_latitude',
    'pickup_longitude',
    'dropoff_longitude',
    'dropoff_latitude'
    
]

df = df.drop(columns=cols_to_drop)
```

```python
df.shape
```

```python
df.columns
```

```python
df.head()
```

```python
# 1. Remove rows with negative or zero duration
df = df[df['duration_minutes'] > 0]

# 2. Remove rows with negative fare amounts (and optionally zero fare if they are free/error trips)
df = df[df['fare_amount'] > 0]

# 3. Filter out unrealistic trip distances (e.g., must be greater than 0 and less than a reasonable maximum like 100 miles)
df = df[(df['trip_distance'] > 0) & (df['trip_distance'] <= 300)]

# Reset index after dropping rows
df.reset_index(drop=True, inplace=True)
```

```python
# Filter out rows where RatecodeID is equal to 99
df = df[df['RatecodeID'] != 99]

# Optional: Reset the index after dropping rows
df.reset_index(drop=True, inplace=True)
```

### Model building

```python
from sklearn.preprocessing import StandardScaler, PowerTransformer

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, root_mean_squared_error, r2_score
```

```python
df.columns
```

```python
num_cols = ['passenger_count', 'trip_distance', 'RatecodeID',
       'duration_minutes', 'day', 'day_of_week', 'is_weekend', 'hour',
       'is_night']
```

```python
num_transformer = Pipeline([('Scaling',StandardScaler()),
                           ['Transformer',PowerTransformer()]])
```

```python
num_transformer
```

```python
preprocessing = ColumnTransformer([('Numeric transformer',num_transformer,num_cols)])
```

```python
preprocessing
```

```python
x = df.drop(columns=['fare_amount'])
y = df['fare_amount']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=23)
```

```python
lr = Pipeline([('Preprocessing',preprocessing),
              ('Model Building',LinearRegression())])
lr.fit(x_train,y_train)
y_pred = lr.predict(x_test)
print('mean absolute error: ',mean_absolute_error(y_test,y_pred))
print('root mean squared error: ',root_mean_squared_error(y_test,y_pred))
print('r2 score: ',r2_score(y_test,y_pred))
```

```python
dt = Pipeline([('Preprocessing',preprocessing),
              ('Model Building',DecisionTreeRegressor())])
dt.fit(x_train,y_train)
y_pred = dt.predict(x_test)
print('mean absolute error: ',mean_absolute_error(y_test,y_pred))
print('root mean squared error: ',root_mean_squared_error(y_test,y_pred))
print('r2 score: ',r2_score(y_test,y_pred))
```

```python
rf = Pipeline([('Preprocessing',preprocessing),
              ('Model Building',RandomForestRegressor())])
rf.fit(x_train,y_train)
y_pred = rf.predict(x_test)
print('mean absolute error: ',mean_absolute_error(y_test,y_pred))
print('root mean squared error: ',root_mean_squared_error(y_test,y_pred))
print('r2 score: ',r2_score(y_test,y_pred))
```

### Hyper Parameter Tuning

```python
import optuna
```

```python
def objective(trial):
    params = {
        'n_estimators': trial.suggest_int('n_estimators', 50, 300, step=25),
        'max_depth': trial.suggest_int('max_depth', 5, 30, log=False),
        'min_samples_split': trial.suggest_int('min_samples_split', 2, 20),
        'min_samples_leaf': trial.suggest_int('min_samples_leaf', 1, 10),
        'max_features': trial.suggest_categorical('max_features', ['sqrt', 'log2', None]),
        'random_state': 42,
        'n_jobs': -1
    }
    
    model = Pipeline([
        ('Data Preprocessing', preprocessing),
        ('Model Building', RandomForestRegressor(**params))
    ])
    
    model.fit(x_train, y_train)
    
    y_pred = model.predict(x_test)
    
    score = r2_score(y_test, y_pred)
    return score

study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=50)

print("Best Hyperparameters:", study.best_params)
print("Best R2 Score:", study.best_value)
```

```python
rf = Pipeline([('Preprocessing',preprocessing),
              ('Model Building',RandomForestRegressor(n_estimators=50,max_depth=12,min_samples_split=19,min_samples_leaf=3,max_features='log2'))])
rf.fit(x_train,y_train)
y_pred = rf.predict(x_test)
print('mean absolute error: ',mean_absolute_error(y_test,y_pred))
print('root mean squared error: ',root_mean_squared_error(y_test,y_pred))
print('r2 score: ',r2_score(y_test,y_pred))
```
